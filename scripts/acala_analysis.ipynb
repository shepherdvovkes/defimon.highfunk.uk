# Acala Network Data Analysis

This notebook provides interactive analysis of the Acala network data collected by the investigation scripts.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
from pathlib import Path

# Set up plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
```

## Load Investigation Data

```python
# Load blocks data
blocks_df = pd.read_csv('acala_data_investigation/acala_blocks_sample.csv')
print(f"Loaded {len(blocks_df)} blocks")
blocks_df.head()
```

```python
# Load extrinsics data
extrinsics_df = pd.read_csv('acala_data_investigation/acala_extrinsics_sample.csv')
print(f"Loaded {len(extrinsics_df)} extrinsics")
extrinsics_df.head()
```

## Load ML Data

```python
# Load ML features
features_df = pd.read_csv('acala_ml_data/acala_features.csv')
print(f"Loaded {len(features_df)} ML feature samples")
features_df.head()
```

```python
# Load ML results
with open('acala_ml_data/acala_ml_results.json', 'r') as f:
    ml_results = json.load(f)

print("ML Model Results:")
for model_name, result in ml_results.items():
    print(f"{model_name}: RMSE = {result['rmse']:.4f}")
```

## Data Visualization

```python
# Block size distribution
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.hist(blocks_df['block_size'], bins=20, alpha=0.7)
plt.title('Block Size Distribution')
plt.xlabel('Block Size')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(blocks_df['extrinsics_count'], bins=20, alpha=0.7)
plt.title('Extrinsics Count Distribution')
plt.xlabel('Extrinsics Count')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
# Feature importance visualization
for model_name, result in ml_results.items():
    importance = result['feature_importance']
    top_features = sorted(importance.items(), key=lambda x: x[1], reverse=True)[:10]
    
    features, scores = zip(*top_features)
    
    plt.figure(figsize=(10, 6))
    plt.barh(range(len(features)), scores)
    plt.yticks(range(len(features)), features)
    plt.xlabel('Feature Importance')
    plt.title(f'Top 10 Feature Importance - {model_name.replace("_", " ").title()}')
    plt.gca().invert_yaxis()
    plt.tight_layout()
    plt.show()
```
